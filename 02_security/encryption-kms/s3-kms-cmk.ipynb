```python
import boto3
```

```python
def create_cmk():
    """
    Creates a Customer Managed Key (CMK) in AWS KMS.
    Returns the ARN of the created CMK.
    """
    client = boto3.client('kms')
    response = client.create_key(
        Description='My CMK for S3 encryption',
        KeyUsage='ENCRYPT_DECRYPT'
    )
    return response['KeyMetadata']['Arn']
```

```python
def create_encrypted_bucket(bucket_name, cmk_arn):
    """
    Creates an S3 bucket and enables server-side encryption using the specified CMK.
    """
    s3 = boto3.client('s3')
    response = s3.create_bucket(
        Bucket=bucket_name
    )
    
    s3.put_bucket_encryption(
        Bucket=bucket_name,
        ServerSideEncryptionConfiguration={
            'Rules': [
                {
                    'ApplyServerSideEncryptionByDefault': {
                        'SSEAlgorithm': 'aws:kms',
                        'KMSMasterKeyID': cmk_arn
                    }
                }
            ]
        }
    )
    return response
```

```python
# Create a CMK in AWS KMS
cmk_arn = create_cmk()
print("Created CMK ARN:", cmk_arn)

# Create an S3 bucket and enable encryption using the created CMK
bucket_name = 'cmk-encrypted-bucket-using-sdk'
create_encrypted_bucket(bucket_name, cmk_arn)
print("Created S3 bucket with encryption using CMK")
```
